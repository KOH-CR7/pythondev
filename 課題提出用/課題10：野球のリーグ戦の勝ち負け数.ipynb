```python
#クラスの作成
class Baseball_Team:
    #インスタンスメンバ変数
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw
    #計算式
    def calc_win_rate(self):
        total_games = self.win + self.lose
        return self.win / total_games

#チーム名＋結果
archers = Baseball_Team("Archers", 77, 64, 2)
archers.calc_win_rate()

meteors = Baseball_Team("Meteors", 71, 69, 3)
meteors. calc_win_rate()

cheetahs = Baseball_Team("Cheetahs", 69, 68, 6)
cheetahs. calc_win_rate()

sharks = Baseball_Team("Sharks", 70, 70, 3)
sharks. calc_win_rate()

pythons = Baseball_Team("Pythons", 68, 73, 2)
pythons. calc_win_rate()

owls = Baseball_Team("Owls", 59, 82, 2)
owls. calc_win_rate()

teams = [archers, meteors, cheetahs, sharks, pythons, owls]

#勝敗数表示
def show_team_result(team_list):
    print("|   team     | win|lose|draw| rate |")
    for team in team_list:
        win_rate = team.calc_win_rate()
        print(f"| {team.name:<10} | {team.win:^2} | {team.lose} | {team.draw:^3}| {win_rate:.<5.3}| ")

show_team_result(teams)
```
