```python
#アルファベットの辞書を作成
import string
character = list(string.ascii_lowercase)
character = {i: 0 for i in character}

#英単語の入力を受付しリスト化
word_list = []
while True:
    word_input = input ("英単語を入力してください：")
    print(f"英単語を入力してください:{word_input}")
    if word_input == "":
        break 
    word_list.append(word_input)
    #英単語の個数をリスト化
    for char in word_input.lower():
        if char in character:
            character[char] +=1

word_list.sort()
print(f"入力した英単語{word_list}")

#文字数カウント
for abc, count in sorted(character.items()):
    count = character[abc]
    if count >0:
        print(f"{abc}が{count}個ありました")

```
