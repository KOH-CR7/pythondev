```python
rent = float(input("賃料を入力してください: "))
common_area_fee = float(input("共益費・管理費を入力してください: "))
parking_fee = float(input("駐車場代を入力してください: "))

rent_difference = rent - 110000

if rent_difference < 0:
    rent_calculation = (110000 * 0.25)
else:
    rent_calculation = rent_difference + (110000 * 0.25)

total_self_payment = rent_calculation + common_area_fee + parking_fee

print(f"自己負担額の合計金額は {total_self_payment:,.0f} 円です。")

current_total = 75500
payment_difference = total_self_payment - current_total

print(f"現在の合計金額（{current_total:,.0f}円）との差額は {payment_difference:,.0f} 円です。")
```
